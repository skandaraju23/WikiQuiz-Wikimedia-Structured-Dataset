```python
!pip install -q datasets pyarrow

from datasets import load_dataset
import pandas as pd

# Load the first 10,000 articles directly into memory
print("⏳ Loading dataset from Hugging Face...")
raw_data = load_dataset("wikimedia/wikipedia", "20231101.en", split="train[:10000]")

# Convert to pandas DataFrame
df = raw_data.to_pandas()

# Align columns with what WikiQuiz requires ('name', 'abstract', 'description')
df["name"] = df["title"]
df["abstract"] = df["text"]
df["description"] = ""

print("✅ Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", df.columns.tolist())
```

```python
# ============================================================
#                         WIKIQUIZ
#          Wikimedia Structured Wikipedia Dataset
#               WikiClub Tech - Open Source Day
# ============================================================

import gradio as gr
import pandas as pd
import random
import re
import json

# ============================================================
# SETTINGS
# ============================================================

MAX_QUESTIONS = 5
QUESTION_TIME = 30


# ============================================================
# CHECK DATASET
# ============================================================

if "df" not in globals():
    raise RuntimeError(
        "❌ Dataset 'df' is not loaded.\n"
        "Please run your dataset-loading cell first."
    )

print("=" * 60)
print("📚 WIKIQUIZ")
print("=" * 60)
print("✅ Dataset loaded")
print("✅ Dataset rows:", len(df))
print("✅ Gradio version:", gr.__version__)
print("✅ Questions:", MAX_QUESTIONS)
print("✅ Timer:", QUESTION_TIME, "seconds")
print("=" * 60)


# ============================================================
# HELPER FUNCTION
# ============================================================

def clean_text(value):

    if value is None:
        return ""

    try:
        if pd.isna(value):
            return ""
    except:
        pass

    return str(value)


def get_col(column_name):

    if column_name in df.columns:

        return (
            df[column_name]
            .fillna("")
            .astype(str)
        )

    return pd.Series(
        [""] * len(df),
        index=df.index
    )


# ============================================================
# SEARCH WIKIMEDIA DATASET
# ============================================================

def search_articles(topic):

    topic = clean_text(topic).strip()

    if not topic:
        return pd.DataFrame()

    topic_lower = topic.lower()

    names = get_col("name")

    abstracts = get_col("abstract")

    descriptions = get_col("description")


    # --------------------------------------------------------
    # 1. EXACT TITLE MATCH
    # --------------------------------------------------------

    exact_mask = (
        names.str.strip().str.lower()
        == topic_lower
    )

    exact_results = df[
        exact_mask
    ].copy()

    if not exact_results.empty:

        exact_results["_length"] = (
            exact_results["abstract"]
            .fillna("")
            .astype(str)
            .str.len()
        )

        exact_results = (
            exact_results
            .sort_values(
                "_length",
                ascending=False
            )
            .drop(
                columns=["_length"]
            )
        )

        return exact_results.head(20)


    # --------------------------------------------------------
    # 2. TITLE STARTS WITH TOPIC
    # --------------------------------------------------------

    starts_mask = names.str.lower().str.startswith(
        topic_lower,
        na=False
    )

    starts_results = df[
        starts_mask
    ].copy()

    if not starts_results.empty:

        starts_results["_length"] = (
            starts_results["abstract"]
            .fillna("")
            .astype(str)
            .str.len()
        )

        starts_results = (
            starts_results
            .sort_values(
                "_length",
                ascending=False
            )
            .drop(
                columns=["_length"]
            )
        )

        return starts_results.head(20)


    # --------------------------------------------------------
    # 3. TITLE CONTAINS TOPIC
    # --------------------------------------------------------

    contains_mask = names.str.lower().str.contains(
        re.escape(topic_lower),
        regex=True,
        na=False
    )

    contains_results = df[
        contains_mask
    ].copy()

    if not contains_results.empty:

        contains_results["_length"] = (
            contains_results["abstract"]
            .fillna("")
            .astype(str)
            .str.len()
        )

        contains_results = (
            contains_results
            .sort_values(
                "_length",
                ascending=False
            )
            .drop(
                columns=["_length"]
            )
        )

        return contains_results.head(20)


    # --------------------------------------------------------
    # 4. SEARCH ARTICLE CONTENT
    # --------------------------------------------------------

    search_text = (
        names.str.lower()
        + " "
        + abstracts.str.lower()
        + " "
        + descriptions.str.lower()
    )

    content_mask = search_text.str.contains(
        re.escape(topic_lower),
        regex=True,
        na=False
    )

    return df[
        content_mask
    ].head(20)


# ============================================================
# GET ARTICLE TEXT
# ============================================================

def get_article_text(row):

    abstract = clean_text(
        row.get(
            "abstract",
            ""
        )
    )

    description = clean_text(
        row.get(
            "description",
            ""
        )
    )

    if len(abstract) >= 100:
        return abstract

    if len(description) >= 100:
        return description

    return (
        abstract
        + " "
        + description
    ).strip()


# ============================================================
# SPLIT SENTENCES
# ============================================================

def get_sentences(text):

    text = clean_text(text)

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    if not text:
        return []

    sentences = re.split(
        r"(?<=[.!?])\s+",
        text
    )

    return [
        sentence.strip()
        for sentence in sentences
        if len(sentence.strip()) >= 35
    ]


# ============================================================
# GENERATE QUESTIONS
# ============================================================

def make_questions(text):

    sentences = get_sentences(text)

    if not sentences:
        return []

    random.shuffle(sentences)

    candidates = []

    used_answers = set()


    # ========================================================
    # METHOD 1: YEARS / NUMBERS
    # ========================================================

    for sentence in sentences:

        numbers = re.findall(
            r"\b(?:1[0-9]{3}|20[0-9]{2}|[0-9]{1,4})\b",
            sentence
        )

        for number in numbers:

            key = number.lower()

            if key in used_answers:
                continue

            used_answers.add(key)

            candidates.append({
                "sentence": sentence,
                "answer": number
            })

            if len(candidates) >= MAX_QUESTIONS:
                break

        if len(candidates) >= MAX_QUESTIONS:
            break


    # ========================================================
    # METHOD 2: IMPORTANT WORDS
    # ========================================================

    if len(candidates) < MAX_QUESTIONS:

        stop_words = {
            "about",
            "after",
            "also",
            "been",
            "before",
            "being",
            "between",
            "during",
            "each",
            "from",
            "have",
            "has",
            "had",
            "into",
            "more",
            "most",
            "other",
            "such",
            "than",
            "that",
            "their",
            "there",
            "these",
            "they",
            "this",
            "those",
            "through",
            "under",
            "used",
            "were",
            "which",
            "while",
            "where",
            "when",
            "with",
            "would",
            "could",
            "should",
            "often",
            "known",
            "including",
            "called",
            "first",
            "second"
        }

        for sentence in sentences:

            words = re.findall(
                r"\b[A-Za-z][A-Za-z'-]{4,}\b",
                sentence
            )

            useful_words = []

            for word in words:

                if (
                    word.lower()
                    not in stop_words
                    and word.lower()
                    not in used_answers
                ):

                    useful_words.append(
                        word
                    )

            if not useful_words:
                continue

            # Pick a meaningful longer word
            useful_words.sort(
                key=len,
                reverse=True
            )

            answer = useful_words[0]

            used_answers.add(
                answer.lower()
            )

            candidates.append({
                "sentence": sentence,
                "answer": answer
            })

            if len(candidates) >= MAX_QUESTIONS:
                break


    # ========================================================
    # CREATE MULTIPLE CHOICE QUESTIONS
    # ========================================================

    questions = []

    all_answers = [
        item["answer"]
        for item in candidates
    ]


    for item in candidates:

        sentence = item["sentence"]

        correct = item["answer"]


        # ----------------------------------------------------
        # CREATE BLANK
        # ----------------------------------------------------

        blank_sentence = re.sub(
            re.escape(correct),
            "_____",
            sentence,
            count=1,
            flags=re.IGNORECASE
        )


        # ----------------------------------------------------
        # WRONG ANSWERS
        # ----------------------------------------------------

        wrong_answers = []

        other_answers = [
            answer
            for answer in all_answers
            if answer.lower()
            != correct.lower()
        ]

        random.shuffle(
            other_answers
        )


        for answer in other_answers:

            if answer not in wrong_answers:

                wrong_answers.append(
                    answer
                )

            if len(wrong_answers) == 3:
                break


        # ----------------------------------------------------
        # FALLBACK ANSWERS
        # ----------------------------------------------------

        fallback = [
            "10",
            "20",
            "25",
            "50",
            "100",
            "200",
            "500",
            "1000",
            "1950",
            "1960",
            "1970",
            "1980",
            "1990",
            "2000",
            "2010",
            "2020"
        ]

        random.shuffle(
            fallback
        )


        for option in fallback:

            if (
                option != correct
                and option not in wrong_answers
            ):

                wrong_answers.append(
                    option
                )

            if len(wrong_answers) == 3:
                break


        # Final safety fallback
        while len(wrong_answers) < 3:

            option = str(
                random.randint(
                    1,
                    999
                )
            )

            if (
                option != correct
                and option not in wrong_answers
            ):

                wrong_answers.append(
                    option
                )


        # ----------------------------------------------------
        # SHUFFLE OPTIONS
        # ----------------------------------------------------

        options = [
            correct,
            wrong_answers[0],
            wrong_answers[1],
            wrong_answers[2]
        ]

        random.shuffle(options)


        correct_letter = "ABCD"[
            options.index(correct)
        ]


        # ----------------------------------------------------
        # QUESTION
        # ----------------------------------------------------

        question_text = (
            "According to the selected "
            "Wikipedia article, which value "
            "or term correctly completes "
            "the statement?\n\n"
            + blank_sentence
        )


        questions.append({

            "question":
                question_text,

            "options":
                options,

            "correct":
                correct,

            "correct_letter":
                correct_letter
        })


    return questions[:MAX_QUESTIONS]


# ============================================================
# DISPLAY QUESTION
# ============================================================

def display_question(
    questions,
    index
):

    if not questions:
        return "No question available."

    if index >= len(questions):
        return "🏆 Quiz Finished"

    q = questions[index]

    output = (
        f"## ❓ Question {index + 1}\n\n"
        f"{q['question']}\n\n"
    )

    for i, option in enumerate(
        q["options"]
    ):

        letter = "ABCD"[i]

        output += (
            f"**{letter}.** {option}\n\n"
        )

    return output


# ============================================================
# START QUIZ
# ============================================================

def start_quiz(topic):

    results = search_articles(
        topic
    )


    if results.empty:

        return (
            "❌ **No article found.**\n\n"
            "Try **India, Space, Tesla, "
            "Car, Cricket, Basketball**.",

            "No question available.",

            None,

            "",

            0,

            0,

            False,

            QUESTION_TIME,

            f"⏱️ **Time: {QUESTION_TIME}s**",

            "No quiz started.",

            "Try another topic."
        )


    # --------------------------------------------------------
    # BEST MATCHING ARTICLE
    # --------------------------------------------------------

    row = results.iloc[0]


    article_name = clean_text(
        row.get(
            "name",
            "Wikipedia Article"
        )
    )


    article_text = get_article_text(
        row
    )


    # --------------------------------------------------------
    # GENERATE QUESTIONS
    # --------------------------------------------------------

    questions = make_questions(
        article_text
    )


    if not questions:

        return (
            f"❌ Article found: "
            f"**{article_name}**, but "
            "there was not enough information "
            "to generate questions.",

            "No question available.",

            None,

            "",

            0,

            0,

            False,

            QUESTION_TIME,

            f"⏱️ **Time: {QUESTION_TIME}s**",

            "Try another topic.",

            "No quiz started."
        )


    # --------------------------------------------------------
    # SAVE QUESTIONS
    # --------------------------------------------------------

    questions_json = json.dumps(
        questions
    )


    # --------------------------------------------------------
    # FIRST QUESTION
    # --------------------------------------------------------

    question_text = display_question(
        questions,
        0
    )


    # --------------------------------------------------------
    # DATASET RESULT
    # --------------------------------------------------------

    dataset_text = (
        "## 📚 Wikimedia Dataset Result\n\n"

        f"**Article:** {article_name}\n\n"

        f"**Questions Generated:** "
        f"{len(questions)}\n\n"

        "**Dataset:** Wikimedia Structured "
        "Wikipedia Dataset\n\n"

        "### Article Information\n\n"

        f"{article_text[:1500]}"
    )


    return (
        dataset_text,

        question_text,

        None,

        questions_json,

        0,

        0,

        False,

        QUESTION_TIME,

        f"⏱️ **Time: {QUESTION_TIME}s**",

        f"Question 1 of {len(questions)}",

        "🎯 Select A, B, C or D and click "
        "**Submit Answer**."
    )


# ============================================================
# SUBMIT ANSWER
# FINAL SCORE IS SHOWN AFTER LAST QUESTION
# ============================================================

def submit_answer(
    selected,
    questions_json,
    index,
    score,
    answered,
    time_left
):

    if not questions_json:

        return (
            "❌ Please start the quiz first.",
            score,
            answered
        )


    questions = json.loads(
        questions_json
    )


    if index >= len(questions):

        return (
            "🏆 Quiz already completed.",
            score,
            True
        )


    if answered:

        return (
            "⚠️ Already answered. "
            "Click **Next Question**.",
            score,
            True
        )


    if not selected:

        return (
            "⚠️ Please select A, B, C or D.",
            score,
            False
        )


    if time_left <= 0:

        return (
            "⏰ Time's up!",
            score,
            True
        )


    q = questions[index]


    correct_letter = q[
        "correct_letter"
    ]

    correct_answer = q[
        "correct"
    ]


    # ========================================================
    # CHECK ANSWER
    # ========================================================

    if selected == correct_letter:

        new_score = score + 1

        answer_message = (
            "## ✅ Correct!\n\n"

            f"Your answer: **{selected}**\n\n"

            f"Correct answer: "
            f"**{correct_letter}. "
            f"{correct_answer}**\n\n"

            f"🎯 Current Score: "
            f"**{new_score}/{len(questions)}**"
        )

    else:

        new_score = score

        answer_message = (
            "## ❌ Incorrect\n\n"

            f"Your answer: **{selected}**\n\n"

            f"Correct answer: "
            f"**{correct_letter}. "
            f"{correct_answer}**\n\n"

            f"🎯 Current Score: "
            f"**{new_score}/{len(questions)}**"
        )


    # ========================================================
    # LAST QUESTION
    # SHOW FINAL SCORE IMMEDIATELY
    # ========================================================

    if index == len(questions) - 1:

        total = len(questions)

        percentage = round(
            (new_score / total) * 100,
            1
        )


        final_message = (
            answer_message

            + "\n\n"
            + "---"
            + "\n\n"

            + "# 🏆 QUIZ COMPLETED!"

            + "\n\n"

            + f"## 🎯 Final Score: "
            f"**{new_score}/{total}**"

            + "\n\n"

            + f"## 📊 Percentage: "
            f"**{percentage}%**"

            + "\n\n"

            + "🎉 Congratulations! "
            "You completed all questions."
        )


        return (
            final_message,
            new_score,
            True
        )


    # ========================================================
    # NORMAL QUESTION
    # ========================================================

    return (
        answer_message,
        new_score,
        True
    )


# ============================================================
# TIMER
# AUTOMATICALLY MOVES TO NEXT QUESTION
# ============================================================

def timer_tick(
    time_left,
    answered,
    questions_json,
    current_index,
    score
):

    # --------------------------------------------------------
    # QUIZ NOT STARTED
    # --------------------------------------------------------

    if not questions_json:

        return (
            QUESTION_TIME,
            answered,
            f"⏱️ **Time: {QUESTION_TIME}s**",
            current_index,
            score,
            gr.skip(),
            gr.skip(),
            gr.skip()
        )


    # --------------------------------------------------------
    # ALREADY ANSWERED
    # --------------------------------------------------------

    if answered:

        if time_left <= 0:

            return (
                0,
                True,
                "⏰ **Time's Up!**",
                current_index,
                score,
                gr.skip(),
                gr.skip(),
                gr.skip()
            )

        return (
            time_left,
            True,
            f"⏱️ **Time: {time_left}s**",
            current_index,
            score,
            gr.skip(),
            gr.skip(),
            gr.skip()
        )


    # --------------------------------------------------------
    # TIME IS UP
    # --------------------------------------------------------

    if time_left <= 1:

        questions = json.loads(
            questions_json
        )

        next_index = (
            current_index + 1
        )


        # ====================================================
        # LAST QUESTION TIMED OUT
        # ====================================================

        if next_index >= len(
            questions
        ):

            total = len(
                questions
            )

            percentage = round(
                (score / total) * 100,
                1
            )


            final_result = (
                "# 🏆 QUIZ COMPLETED!\n\n"

                "⏰ The last question "
                "timed out.\n\n"

                f"## 🎯 Final Score: "
                f"**{score}/{total}**\n\n"

                f"## 📊 Percentage: "
                f"**{percentage}%**\n\n"

                "🎉 You completed WikiQuiz!"
            )


            return (
                0,
                True,
                "🏆 **Quiz Finished**",
                next_index,
                score,
                "## 🎉 All questions completed!",
                None,
                final_result
            )


        # ====================================================
        # AUTOMATIC NEXT QUESTION
        # ====================================================

        next_question_text = display_question(
            questions,
            next_index
        )


        return (
            QUESTION_TIME,

            False,

            f"⏱️ **Time: {QUESTION_TIME}s**",

            next_index,

            score,

            next_question_text,

            None,

            (
                f"⏰ Time's up! "
                f"Moving to Question "
                f"{next_index + 1}..."
            )
        )


    # --------------------------------------------------------
    # NORMAL COUNTDOWN
    # --------------------------------------------------------

    new_time = (
        time_left - 1
    )


    return (
        new_time,

        False,

        f"⏱️ **Time: {new_time}s**",

        current_index,

        score,

        gr.skip(),

        gr.skip(),

        gr.skip()
    )


# ============================================================
# MANUAL NEXT QUESTION
# ============================================================

def next_question(
    questions_json,
    index,
    score,
    answered,
    time_left
):

    if not questions_json:

        return (
            "❌ Please start a quiz first.",
            None,
            index,
            score,
            answered,
            time_left,
            f"⏱️ Time: {time_left}s",
            "No quiz started.",
            "No result yet.",
            questions_json
        )


    questions = json.loads(
        questions_json
    )


    if not answered:

        return (
            "⚠️ Submit your answer first.",
            None,
            index,
            score,
            answered,
            time_left,
            f"⏱️ Time: {time_left}s",
            f"Question {index + 1} "
            f"of {len(questions)}",
            "Please answer this question.",
            questions_json
        )


    next_index = (
        index + 1
    )


    # --------------------------------------------------------
    # QUIZ FINISHED
    # --------------------------------------------------------

    if next_index >= len(
        questions
    ):

        total = len(
            questions
        )

        percentage = round(
            (score / total) * 100,
            1
        )


        final_result = (
            "# 🏆 QUIZ COMPLETED!\n\n"

            f"## 🎯 Final Score: "
            f"**{score}/{total}**\n\n"

            f"## 📊 Percentage: "
            f"**{percentage}%**\n\n"

            "🎉 You completed WikiQuiz!"
        )


        return (
            "## 🎉 All questions completed!",

            None,

            next_index,

            score,

            True,

            0,

            "🏆 **Quiz Finished**",

            "Quiz Completed",

            final_result,

            "FINISHED"
        )


    # --------------------------------------------------------
    # SHOW NEXT QUESTION
    # --------------------------------------------------------

    question_text = display_question(
        questions,
        next_index
    )


    return (
        question_text,

        None,

        next_index,

        score,

        False,

        QUESTION_TIME,

        f"⏱️ **Time: {QUESTION_TIME}s**",

        f"Question {next_index + 1} "
        f"of {len(questions)}",

        "Select your answer.",

        questions_json
    )


# ============================================================
# NEW QUIZ
# ============================================================

def reset_quiz():

    return (
        "",

        "Enter a topic and click "
        "**Generate Quiz**.",

        None,

        "",

        0,

        0,

        False,

        QUESTION_TIME,

        f"⏱️ **Time: {QUESTION_TIME}s**",

        "Ready for a new quiz.",

        "🎯 Welcome to WikiQuiz!"
    )


# ============================================================
# GRADIO INTERFACE
# ============================================================

with gr.Blocks(
    title="WikiQuiz - Wikimedia Dataset"
) as app:


    # ========================================================
    # HEADER
    # ========================================================

    gr.HTML(
        """
        <div style="
            padding:22px;
            border-radius:15px;
            text-align:center;
            background:linear-gradient(
                90deg,#202020,#444444
            );
            margin-bottom:15px;
        ">

            <h1 style="
                color:white;
                margin:0;
            ">
                📚 WikiQuiz
            </h1>

            <p style="
                color:#dddddd;
                margin:8px;
            ">
                Wikimedia Structured Wikipedia Dataset
            </p>

            <p style="
                color:#bbbbbb;
                margin:5px;
            ">
                WikiClub Tech • Open Source Day
            </p>

        </div>
        """
    )


    # ========================================================
    # SEARCH AREA
    # ========================================================

    with gr.Row():

        topic = gr.Textbox(
            label="🔎 Search Topic",
            placeholder=(
                "Try: India, Space, Tesla, "
                "Car, Cricket, Basketball"
            ),
            scale=4
        )


        generate = gr.Button(
            "🎯 Generate Quiz",
            variant="primary",
            scale=1
        )


        new_quiz = gr.Button(
            "🔄 New Quiz",
            scale=1
        )


    # ========================================================
    # DATASET RESULT
    # ========================================================

    dataset_result = gr.Markdown(
        "Enter a topic and click "
        "**Generate Quiz**."
    )


    # ========================================================
    # PROGRESS + TIMER
    # ========================================================

    with gr.Row():

        progress = gr.Markdown(
            "Ready for a new quiz."
        )


        timer_display = gr.Markdown(
            f"⏱️ **Time: {QUESTION_TIME}s**"
        )


    # ========================================================
    # QUESTION
    # ========================================================

    question = gr.Markdown(
        "Your question will appear here."
    )


    # ========================================================
    # ANSWER
    # ========================================================

    answer = gr.Dropdown(
        choices=[
            "A",
            "B",
            "C",
            "D"
        ],
        label="Choose your answer",
        value=None
    )


    # ========================================================
    # BUTTONS
    # ========================================================

    with gr.Row():

        submit = gr.Button(
            "✅ Submit Answer",
            variant="primary"
        )


        next_button = gr.Button(
            "➡️ Next Question"
        )


    # ========================================================
    # RESULT
    # ========================================================

    result = gr.Markdown(
        "🎯 Quiz result will appear here."
    )


    # ========================================================
    # SESSION STATES
    # ========================================================

    questions_state = gr.State("")

    index_state = gr.State(0)

    score_state = gr.State(0)

    answered_state = gr.State(False)

    time_state = gr.State(
        QUESTION_TIME
    )


    # ========================================================
    # GENERATE QUIZ
    # ========================================================

    generate.click(

        fn=start_quiz,

        inputs=[
            topic
        ],

        outputs=[
            dataset_result,
            question,
            answer,
            questions_state,
            index_state,
            score_state,
            answered_state,
            time_state,
            timer_display,
            progress,
            result
        ]
    )


    # ========================================================
    # SUBMIT ANSWER
    # ========================================================

    submit.click(

        fn=submit_answer,

        inputs=[
            answer,
            questions_state,
            index_state,
            score_state,
            answered_state,
            time_state
        ],

        outputs=[
            result,
            score_state,
            answered_state
        ]
    )


    # ========================================================
    # NEXT QUESTION
    # ========================================================

    next_button.click(

        fn=next_question,

        inputs=[
            questions_state,
            index_state,
            score_state,
            answered_state,
            time_state
        ],

        outputs=[
            question,
            answer,
            index_state,
            score_state,
            answered_state,
            time_state,
            timer_display,
            progress,
            result,
            questions_state
        ]
    )


    # ========================================================
    # NEW QUIZ
    # ========================================================

    new_quiz.click(

        fn=reset_quiz,

        inputs=[],

        outputs=[
            dataset_result,
            question,
            answer,
            questions_state,
            index_state,
            score_state,
            answered_state,
            time_state,
            timer_display,
            progress,
            result
        ]
    )


    # ========================================================
    # TIMER
    # ========================================================

    quiz_timer = gr.Timer(
        value=1,
        active=True
    )


    quiz_timer.tick(

        fn=timer_tick,

        inputs=[
            time_state,
            answered_state,
            questions_state,
            index_state,
            score_state
        ],

        outputs=[
            time_state,
            answered_state,
            timer_display,
            index_state,
            score_state,
            question,
            answer,
            result
        ],

        queue=False
    )


# ============================================================
# LAUNCH
# ============================================================

print()
print("=" * 60)
print("🚀 WIKIQUIZ READY")
print("=" * 60)
print("Dataset rows:", len(df))
print("Gradio:", gr.__version__)
print("Maximum questions:", MAX_QUESTIONS)
print("Time per question:", QUESTION_TIME, "seconds")
print("=" * 60)

app.launch(
    share=True
)
```
